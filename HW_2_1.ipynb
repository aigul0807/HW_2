```python
import numpy as np
import sys
import array

## 1. Какие ещё существуют коды типов?
a1 = array.array('i', [1, 2, 3]) # -- целые числа, которые могут быть как положительными, так и отрицательными
print(sys.getsizeof(a1))

a2 = array.array('b', [122, 24]) # -- целые числа без знака от 0 до 255
print(sys.getsizeof(a2))

a3 = array.array('B', [True, False]) # -- Булевы (логические значения True или False)
print(sys.getsizeof(a3))

a4 = array.array('u', ['5', 'L']) # -- символы Unicode
print(sys.getsizeof(a4))

a5 = array.array('f', [-7.676, 0.1]) # -- числа с плавающей точкой (плавающие точки одинарной точности)
print(sys.getsizeof(a5))

a6 = array.array('d', [0.6263546725467253745723, 9.8726365735545537632]) # -- числа с двойной точностью (плавающие точки двойной точности)
print(a6)
print(sys.getsizeof(a6))

a7 = array.array('l', [-8326498264286754, 28367482638]) # -- длинные целые числа от -(2^31) до (2^31 - 1)
print(sys.getsizeof(a7))

a8 = array.array('L', [3926928639823, 2937825834872093]) # -- длинные целые числа без знака от 0 до (2^32 - 1)
print(sys.getsizeof(a8))

a9 = array.array('q', [-234567890987, 45678876545678]) # -- длинные целые числа от -(2^63) до (2^63 - 1)
print(sys.getsizeof(a9))

a10 = array.array('Q', [0, 7123571537512712537]) # -- длинные целые числа без знака от 0 до (2^63 - 1)
print(sys.getsizeof(a10))
```

```python
## 7. Написать код для создания срезов массива 3 на 4
a = np.random.randint(0, 10, (3, 4))
print(a)
print('_______________')

# - первые две строки и три столбца
print(a[:2, :3])
print('_______________')

# - первые три строки и второй столбец
print(a[:3, 1:2])
print('_______________')

# - все строки и столбцы в обратном порядке
print(a[::-1, ::-1])
print('_______________')

# - второй столбец
print(a[:, 1])
print('_______________')

# - третья строка
print(a[2, :])
```

```python
# 9. Продемонстрировать использование newaxis для получения вектора-столбца и вектора-строки

arr = np.array([1, 2, 3])

# Получение вектора-столбца
col_vect = arr[:, np.newaxis]
print(col_vect)

# Получение вектора-строки
row_vect = arr[np.newaxis, :]
print(row_vect)
```

```python
# 10. Разберитесь, как работает метод dstack

a1 = np.array([[1, 2], [3, 4]])
a2 = np.array([[5, 6], [7, 8]])

# Стекование массивов вдоль третьей оси
a = np.dstack((a1, a2))
print(a)

# Входные массивы: dstack() принимает последовательность массивов, которые должны иметь одинаковое количество измерений и одинаковый размер вдоль всех осей, кроме последней. Например, если у вас есть два двумерных массива, они должны иметь одинаковое количество строк, но могут иметь разное количество столбцов.
# Стекование: метод dstack() объединяет массивы вдоль третьей оси (глубинной оси). Каждый массив в последовательности становится новым срезом вдоль третьей оси в результирующем массиве. Если исходные массивы двумерные, то результирующий массив будет трехмерным.
# Форма результирующего массива: форма результирующего массива будет равна форме входных массивов, но с добавлением новой оси в конце. Например, если у вас есть два двумерных массива с формой (2, 3), то результирующий массив после dstack() будет иметь форму (2, 3, 2).
```

```python
# 11. Разберитесь, как работают методы split, vsplit, hsplit, dsplit

a = np.arange(16).reshape(4, 4)
print(a)
print('______________')

# Разделение массива на 4 подмассива вдоль оси 0 и оси 1, соответственно
a1 = np.split(a, 4, axis=0)
print(a1)
print('______________')

a2 = np.split(a, 4, axis=1)
print(a2)
print('______________')

# Разделение массива на 2 подмассива вдоль оси 0 с помощью vsplit
a3 = np.vsplit(a, 2)
print(a3)
print('______________')

# Разделение массива на 2 подмассива вдоль оси 1 с помощью hsplit
a4 = np.hsplit(a, 2)
print(a4)
```
