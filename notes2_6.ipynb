**Визуализация данных в Matplotlib**



```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(1)
data = rng.normal(size=1000)

plt.hist(data,
         bins=30, # количество "кусочков"
         density=True,
         alpha=0.5, # прозрачность
         histtype='step', # step - просто контур
         edgecolor='red'
         )
```

```python
x1 = rng.normal(0, 0.8, 1000)
x2 = rng.normal(-2, 1, 1000)
x3 = rng.normal(3, 2, 1000)
args = dict(
     alpha=0.3
 )

plt.hist(x1, **args)
plt.hist(x2, **args)
plt.hist(x3, **args)

plt.show()
```

```python
print(np.histogram(x1, bins=1))
print(np.histogram(x1, bins=2))
print(np.histogram(x1, bins=40))
```

**Двумерные гистограммы**

```python
mean = [0,0]
cov = [[1,1], [1,2]]
```

```python
x, y = rng.multivariate_normal(mean, cov, 10000).T
```

```python
plt.hist2d(x, y, bins=30)
plt.hexbin(x, y, gridsize=30) # шестиугольники

cb = plt.colorbar()

cb.set_label('Points in interval')

print(np.histogram2d(x, y, bins=1))
print(np.histogram2d(x, y, bins=10))
```

**Легенда**

```python
x = np.linspace(0,10,1000)

fig, ax = plt.subplots()

y = np.sin(x[:, np.newaxis] + np.pi * np.arange(0, 2, 0.5))

plt.plot(x, y)

lines = plt.plot(x, y) # plt.Line2d

plt.legend(lines, ['1', 'второй', 'third', '4-й'], loc='upper left')

plt.legend(lines[:2], ['1', '2'])


ax.plot(x, np.sin(x), label='Синус')
ax.plot(x, np.cos(x), label='Косинус')
ax.plot(x, np.cos(x) + 2)
ax.axis('equal')

ax.legend(
     frameon=False, # рамка вокруг легенды
     fancybox=True,
     shadow=True
 )
```

```python
cities = pd.read_csv('/content/california_cities.csv')

lat, lon, pop, area = cities['latd'], cities['longd'], \
    cities['population_total'], cities['area_total_km2']

plt.scatter(lon, lat, c=np.log10(pop), s=area) # s == size, c == color

plt.xlabel('Широта')
plt.ylabel('Долгота')
plt.colorbar()
plt.clim(3, 7) # пределы colorbar

plt.scatter([],[],c='k', alpha=0.5, s=100, label='100 $km^2$')
plt.scatter([],[],c='k', alpha=0.5, s=300, label='300 $km^2$')
plt.scatter([],[],c='k', alpha=0.5, s=500, label='500 $km^2$')

plt.legend(labelspacing=2, frameon=False)

fig, ax = plt.subplots()
lines = []
styles = ['-', '--', '-.', ':']
x = np.linspace(0, 10, 1000)
for i in range(4):
    lines += ax.plot(
        x,
        np.sin(x - i + np.pi / 2),
        styles[i]
    )

ax.axis('equal')

ax.legend(lines[:2], ['line 1', 'line 2'], loc='upper right')

leg = mpl.legend.Legend(ax, lines[1:], ['line 2', 'line 3', 'line 4'], loc='lower left')

ax.add_artist(leg)
```

**Шкалы**

```python
x = np.linspace(0,10,1000)
y = np.sin(x) * np.cos(x[:, np.newaxis])
```

Карты цветов:
1. последовательные (градиент)



```python
plt.imshow(y, cmap='binary')
plt.imshow(y, cmap='viridis')
```

2. дивергентные (два цвета)

```python
plt.imshow(y, cmap='RdBu')
plt.imshow(y, cmap='PuOr')
```

3. качественные (смешиваются без четкого порядка)

```python
plt.imshow(y, cmap='rainbow')
plt.imshow(y, cmap='jet')
```

```python
plt.figure()
plt.subplot(1, 2, 1)
plt.imshow(y, cmap='viridis')
plt.colorbar()
```

```python
plt.subplot(1, 2, 2)
plt.imshow(y, cmap=plt.cm.get_cmap('viridis', 6)) # дискретное деление
plt.colorbar()
plt.clim(-0.25, 0.25)
```
