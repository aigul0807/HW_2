```python
import numpy as np
import pandas as pd

# 1. Привести различные способы создания объектов типа Series
# Для создания Series можно использовать

# - списки Python или массивы NumPy
data = [10, 20, 30, 40, 50]
s = pd.Series(data)
print(s)

print('_______________')

data_array = np.array([10, 20, 30, 40, 50])
s = pd.Series(data_array)
print(s)

print('_______________')

# - скалярные значения
num = 4
s = pd.Series(num, index=[0, 1, 2, 3, 4])
print(s)

print('_______________')

# - словари
data_dict = {"a": 1, "b": 2, "c": 3}
s = pd.Series(data_dict)
print(s)
```

```python
# 2. Привести различные способы создания объектов типа DataFrame
# DataFrame. Способы создания:

# - через объекты Series
s1 = pd.Series([1, 2, 3], name="col1")
s2 = pd.Series([4, 5, 6], name="col2")
df = pd.DataFrame({"col1": s1, "col2": s2})
print(df)

# - через списки словарей
data = [{"name": "A", "age": 25}, {"name": "S", "age": 30}]
df = pd.DataFrame(data)
print(df)


# - через словари объектов Series
data_dict = {"name": pd.Series(["A", "B"]), "age": pd.Series([25, 30])}
df = pd.DataFrame(data_dict)
print(df)


# - через двумерный массив NumPy
data_array = np.array([[2002, 12, 10], [2005, 8, 7]])
df = pd.DataFrame(data_array, columns=["year", "month", "day"])
print(df)


# - через структурированный массив NumPy
data_structured = np.array([(1, "A"), (2, "B")], dtype=[("id", int), ("name", "U10")])
df = pd.DataFrame(data_structured)
print(df)
```

```python
# 3. Объедините два объекта Series с неодинаковыми множествами ключей (индексов) так, чтобы вместо NaN было установлено значение 1
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s2 = pd.Series([4, 5], index=["c", "d"])

print(s1)
print(s2)
print(s1 + s2)

# Объединение Series с заполнением отсутствующих значений значением 1
res = (s1 + s2).fillna(1)

print(res)
```

```python
# 4. Переписать пример с транслированием для DataFrame так, чтоюы вычитание происходило по столбцам

rng = np.random.default_rng()

A = rng.integers(0, 10, (3, 4))
print(A)

df = pd.DataFrame(A, columns=['a', 'b', 'c', 'd'])
print(df)

print(df.T.iloc[0])

print((df.T - df.T.iloc[0]).T)

print(df.T.iloc[0, ::2])

print((df.T - df.T.iloc[0, ::2]).T)
```

```python
# 5. На примере объектов DataFrame продемонстрируйте методов ffill() и bfill()
rng = np.random.default_rng()
data = {'A': [1, 2, np.nan, 4, 5],
        'B': [np.nan, 2, 3, np.nan, 5],
        'C': [1, 2, 3, 4, 5]}
df = pd.DataFrame(data)
print(df)

# Заполнение пропущенных значений методом ffill() -- заполняет пропущенные значения, двигаясь вперед по данным
print(df.ffill())

# Заполнение пропущенных значений методом bfill() -- заполняет пропущенные значения, двигаясь назад по данным
print(df.bfill())
```
