```python
import numpy as np
import pandas as pd

# Pandas -- расширение NumPy (структурированные массивы). Строки и столбцы индексируются метками,  не только числовыми значениями

# Series, DataFrame, Index -- три основным структуры

## Series

data = pd.Series([0.25, 0.5, 0.75, 1.0])
print(data)
print(type(data))

print(data.values)
print(type(data.values))

print(data.index)
print(type(data.index))
```

```python
## DaraFrame -- двумерный массив с явно определёнными индексами. Последовательность "согласованных" объектов Series

population_dict = {
    'city_1': 1001,
    'city_2': 1002,
    'city_3': 1003,
    'city_4': 1004,
    'city_5': 1005
}

area_dict = {
    'city_1': 9991,
    'city_2': 9992,
    'city_3': 9993,
    'city_4': 9994,
    'city_5': 9995
}

population = pd.Series(population_dict)
area = pd.Series(area_dict)

print(population)
print(area)

states = pd.DataFrame({
    'population1': population,
    'area1' : area,
})

print(states)

print(states.values)
print(states.index)
print(states.columns)

print(type(states.values))
print(type(states.index))
print(type(states.columns))
```

```python
print(states['area1'])

# DataFrame. Способы создания
# - через объекты Series
# - через списки словарей
# - через словари объектов Series
# - через двумерный массив NumPy
# - через структурированный массив NumPy
# 2. Привести различные способы создания объектов типа DataFrame
```

```python
# Index -- способ организации ссылки на данные объектов Series и DataFrame. Index -- неизменяем, упорядочен, является мультимножеством(в нём могут быть одинаковые элементы)

ind = pd.Index([2, 3, 5, 7, 11])
print(ind[1])
print(ind[::2])

# ind[1] = 5
```

```python
# Index -- следует соглашениям объекта set (Python)

indA = pd.Index([1, 2, 3, 4, 5])
indB = pd.Index([2, 3, 4, 5, 6])

print(indA.intersection(indB))
```

```python
# Выборка данных из Series
## как словарь

data = pd.Series([0.25, 0.5, 0.75, 1.0], index = ['a', 'b', 'c', 'd'])

print('a' in data)
print('z' in data)

print(data.keys())

print(list(data.items()))

data['a'] = 100
data['z'] = 1000

print(data)
```

```python
## как одномерный массив
data = pd.Series([0.25, 0.5, 0.75, 1.0], index = ['a', 'b', 'c', 'd'])

print(data['a':'c'])
print(data[0:2])
print(data[(data > 0.5) & (data < 1)])
print(data[['a', 'd']])
```

```python
# атрибуты-индексаторы
data = pd.Series([0.25, 0.5, 0.75, 1.0], index = [1, 3, 10, 15])

print(data[1])

print(data.loc[1])
print(data.iloc[1])
```

```python
# Выборка данных из DataFrame
## как словарь

population_dict = {
    'city_1': 1001,
    'city_2': 1002,
    'city_3': 1003,
    'city_4': 1004,
    'city_5': 1005
}

area_dict = {
    'city_1': 9991,
    'city_2': 9992,
    'city_3': 9993,
    'city_4': 9994,
    'city_5': 9995
}

pop = pd.Series(
    {
        'city_1': 1001,
        'city_2': 1002,
        'city_3': 1003,
        'city_4': 1004,
        'city_5': 1005,
    })


area = pd.Series(
    {
        'city_1': 9991,
        'city_2': 9992,
        'city_3': 9993,
        'city_4': 9994,
        'city_5': 9995,
    })
data = pd.DataFrame({'area1':area, 'pop1':pop, 'pop':pop})

print(data)

print(data['area1'])
print(data.area1)

print(data.pop1 is data['pop1'])
print(data.pop is data['pop'])
```

```python
data['new'] = data['area1']

data['new1'] = data['area1'] / data['pop1']

print(data)
```

```python
# двумерный NumPy-массив

data = pd.DataFrame({'area1':area, 'pop1':pop})

print(data)

print(data.values)

print(data.T)

print(data['area1']) # образаемся к столбцу

print(data.values[0]) # обращаемся к строке

print(data.values[0:3])
```

```python
# атрибуты-индексаторы

data = pd.DataFrame({'area1':area, 'pop1':pop, 'pop':pop})

print(data)
print(data.iloc[:3, 1:2])
print(data.loc[:'city_3', 'pop1':'pop'])
print(data.loc[data['pop'] > 1002, ['area1', 'pop']])

data.iloc[0, 2] = 999999

print(data)
```

```python
# NaN -- Not a Number
# NA-значения: NaN, null

# Pandas. Два способа хранения отстутсвующих значений
# 1. индикаторы NaN, None
# 2. null

# None -- объект (накладные расходы). Не работает с sum, min

val1 = np.array([1, 2, 3])
print(val1.sum())

# val1 = np.array([1, None, 2, 3])
# print(val1.sum())

val1 = np.array([1, np.nan, 2, 3])
print(val1)
print(val1.sum())
print(np.sum(val1))
print(np.nansum(val1))
```

```python
x = pd.Series(range(10), dtype = int)
print(x)

x[0] = None
x[1] = np.nan

print(x)

x1 = pd.Series(['a', 'b', 'c'])
print(x1)

x1[0] = None
x1[1] = np.nan

print(x1)
```

```python
#

x2 = pd.Series([1, 2, 3, np.nan, None, pd.NA])
print(x2)

x3 = pd.Series([1, 2, 3, np.nan, None, pd.NA], dtype = 'Int32')
print(x3)

print(x3.isnull())
print(x3[x3.isnull()])
print(x3[x3.notnull()])

print(x3.dropna())
```

```python
df = pd.DataFrame(
    [
        [1, 2, 3, np.nan, None, pd.NA],
        [1, 2, 3, 4, 5, 6],
        [1, np.nan, 3, 4, np.nan, 6]
    ]
)

print(df)
print(df.dropna())
print(df.dropna(axis=0))
print(df.dropna(axis=1))
```

```python
# how
# - all -- все значения NA
# - any -- хотя бы одно значение
# - thresh = x, остаётся, если присутствует МИНИМУМ х НЕПУСТЫХ значений

df = pd.DataFrame(
    [
        [1, 2, 3, np.nan, None, pd.NA],
        [1, 2, 3, None, 5, 6],
        [1, np.nan, 3, None, np.nan, 6]
    ]
)

print(df)
print(df.dropna(axis=1, how = 'all'))
print(df.dropna(axis=1, how = 'any'))
print(df.dropna(axis=1, thresh = 2))
```
