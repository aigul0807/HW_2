```python
import numpy as np

a = np.ones((3, 2))
b = np.arange(3)

# 2 (3, 2)    (3, 2) -> (3, 2)
# 1 (3,)   -> (1, 3) -> (3, 3)

# c = a + b

## Q1: Что надо изменить в последнем примере, чтобы он заработал без ошибок

b = b.reshape((3, 1))  # Изменяем размерность b
c = a + b

print(c)
```
